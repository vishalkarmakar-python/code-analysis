```python
import json
import requests

# Define the Ollama API endpoint
url = "http://localhost:11434/api/generate"

# Define the payload
# Note: Ensure the model "llama3.2:3b" is available in your Ollama instance.
# If not, replace it with a model you have, e.g., "llama3:latest", "mistral", etc.
payload = {
    "model": "llama3.2:3b",
    "prompt": "Why is the sky blue?",
    "stream": False,  # Set to True if you want to stream the response
}

# Set the headers
headers = {"Content-Type": "application/json"}

print(f"Attempting to send request to: {url}")
print(f"Payload: {json.dumps(payload, indent=2)}")
print("-" * 30)

try:
    # Send the POST request
    response = requests.post(url, data=json.dumps(payload), headers=headers, timeout=90)  # Increased timeout

    # Check if the request was successful
    response.raise_for_status()  # This will raise an HTTPError for bad responses (4XX or 5XX)

    # Parse the JSON response
    response_data = response.json()

    print("Request Successful!")
    print("-" * 30)
    print("Full JSON Response:")
    print(json.dumps(response_data, indent=2))

    # Extract and print just the 'response' field if it exists
    if "response" in response_data:
        print("-" * 30)
        print("Model's Answer:")
        print(response_data["response"].strip())
    else:
        print("-" * 30)
        print("The key 'response' was not found in the JSON output.")
        print(
            "This might indicate an issue with the model or the request structure if an answer was expected."
        )


except requests.exceptions.ConnectionError as e:
    print(f"\nError: Could not connect to the Ollama server at {url}.")
    print("Please ensure the Ollama server is running and accessible.")
    print(f"Details: {e}")
except requests.exceptions.Timeout as e:
    print(f"\nError: The request timed out while waiting for a response from {url}.")
    print("The model might be taking too long to generate a response, or the server might be overloaded.")
    print(f"Details: {e}")
except requests.exceptions.HTTPError as e:
    print(f"\nError: HTTP error occurred: {e.response.status_code} {e.response.reason}")
    print("Response Body (if any):")
    try:
        print(json.dumps(e.response.json(), indent=2))
    except json.JSONDecodeError:
        print(e.response.text)
except requests.exceptions.RequestException as e:
    print(f"\nAn unexpected error occurred: {e}")
except json.JSONDecodeError:
    print("\nError: Could not decode the JSON response from the server.")
    print("Raw response content:")
    print(response.text)
```
